```sql
--Registros por Dia Bronce
SELECT 
    date_format(try_to_timestamp(date), 'yyyy-MM-dd') AS fecha,
    COUNT(*) AS total_registros_Bronze
FROM trx_mifare.bronze_uses
WHERE try_to_timestamp(date) >= '2026-11-23'
GROUP BY fecha
ORDER BY fecha ASC;
--Registros por Dia silver
select date_format(try_to_timestamp(trx_date), 'yyyy-MM-dd') AS fecha,
       count(*) As total_registros_Silver 
from trx_mifare.silver_uses
where trx_date >= '2026-11-23'
group by Fecha
order by Fecha ASC
;
--Registros por Dia Gold
select  date_format(try_to_timestamp(Fecha_trx), 'yyyy-MM-dd') AS fecha,
    sum(cant_trx) As total_registros_Gold
from trx_mifare.gold_uses_tx
where Fecha_trx >= '2026-11-23'
group by Fecha
order by Fecha ASC
;
```
