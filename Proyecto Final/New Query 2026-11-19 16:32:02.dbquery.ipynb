```sql
SELECT
    date_format(a.trx_date , 'yyyy-MM-dd')    AS Fecha_trx,
    date_format(a.trx_date, 'yyyy')          AS ano_trx,
    date_format(a.trx_date, 'MM')            AS mes_trx,
    date_format(a.trx_date, 'dd')            AS dia_trx,
    date_format(a.trx_date, 'HH')             AS hora_trx,
    a.line_id                                 AS linea,
    b.ld_id_name                              AS linea_nombre,
    a.tp_id                                   AS empresa,
    d.tp_id_name                              AS empresa_nombre,
    a.app_id                                  AS aplicacion,
    c.app_name                                AS aplicacion_nombre,
    a.use_type                                AS tipo_uso,
    sum(a.fee*10)                                AS total_trx,
    COUNT(*)                               AS cant_trx
FROM trx_mifare.silver_uses a
INNER JOIN trx_mifare.silver_lines b
    ON b.ld_id = a.line_id
INNER JOIN trx_mifare.silver_applications c
    ON c.app_id = a.app_id
INNER JOIN trx_mifare.silver_providers d
    ON d.tp_id = a.tp_id
GROUP BY ALL
ORDER BY Fecha_trx, hora_trx asc
;


select * from trx_mifare.silver_uses;
```
