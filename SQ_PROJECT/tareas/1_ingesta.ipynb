API de OpenSky → RAW (JSON en el Volumen) → bronze (Auto Loader).
Parámetros del Job: `dias`, `start_date`, `aeropuertos`.

```python
%run ../00_Setup
```

```python
dbutils.widgets.text("dias", "3")
dbutils.widgets.text("start_date", "")
dbutils.widgets.text("aeropuertos", "")

dias   = int(dbutils.widgets.get("dias"))
txt    = dbutils.widgets.get("start_date").strip()
inicio = date.fromisoformat(txt) if txt else date.today() - timedelta(days=dias + 1)
lista  = [a.strip().upper() for a in dbutils.widgets.get("aeropuertos").split(",") if a.strip()]

print(f"Parámetros: dias = {dias} | inicio = {inicio} | aeropuertos = {lista or AEROPUERTOS}")
```

```python
download_days(inicio, dias, lista or None)   # API → RAW (idempotente)
run_bronze()                                  # RAW → bronze (Auto Loader)
```
