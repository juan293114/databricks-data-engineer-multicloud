# D_Gold - Modelo estrella

Gold consume las tablas Silver registradas en Unity Catalog y construye
4 dimensiones y 1 tabla de hechos, listas para Power BI.


```python
from pyspark.sql import functions as F

CATALOG = "workspace"
SILVER_SCHEMA = f"{CATALOG}.silver"
GOLD_SCHEMA = f"{CATALOG}.gold"

spark.sql(f"CREATE SCHEMA IF NOT EXISTS {GOLD_SCHEMA}")

stores = spark.table(f"{SILVER_SCHEMA}.stores")



def write_gold(df, table_name):
    (
        df.write
        .format("delta")
        .mode("overwrite")
        .option("overwriteSchema", "true")
        .saveAsTable(table_name)
    )


# Dimensión tiendas
df_dim_stores = (
    stores
    .withColumn("ingestion_timestamp", F.current_timestamp())
    .drop("city")
)
write_gold(df_dim_stores, f"{GOLD_SCHEMA}.dim_stores")


print("GOLD - tabla dimensional dim_stores generado correctamente")
print("Dimensiones: dim_stores")
```
