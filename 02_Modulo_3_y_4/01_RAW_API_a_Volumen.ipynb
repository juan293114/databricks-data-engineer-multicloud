**Objetivo:** traer datos reales y dejarlos **sin tocar** en un Volumen, un archivo JSON por día.

```
GET https://api.tvmaze.com/schedule/web?date=AAAA-MM-DD  ──►  /Volumes/.../raw/schedule/schedule_AAAA-MM-DD.json
```

| Paso | Qué haces |
|---|---|

> Es la misma fuente del laboratorio de los módulos 1 y 2. Aquí cada día es **un archivo nuevo en la carpeta**: eso es lo que Auto Loader va a detectar.

## Paso 1 · Probar la API con un día

````python
r = requests.get("https://api.tvmaze.com/schedule/web", params={"date": "2024-03-01"}, timeout=30)
episodios = r.json()

print("status   :", r.status_code)
print("episodios:", len(episodios))
print("campos   :", list(episodios[0].keys()))
````

👀 **Qué observar:** cada día son cientos de episodios y `_embedded` trae la serie anidada. Es JSON crudo.

## Paso 2 · Parámetros con widgets

````python
dbutils.widgets.text("start_date", "2024-03-01", "Fecha inicial (AAAA-MM-DD)")
dbutils.widgets.text("dias", "5", "Días a descargar")
````

## Paso 3 · Descargar a RAW

Dos reglas de RAW: **no se transforma** y la descarga es **idempotente** (si el archivo existe, no se baja de nuevo).

````python
inicio = date.fromisoformat(dbutils.widgets.get("start_date"))
dias   = int(dbutils.widgets.get("dias"))

download_days(inicio, dias)
````

🧪 **Pruébalo:** ejecuta la celda otra vez. Todos aparecen como "Ya existían".
